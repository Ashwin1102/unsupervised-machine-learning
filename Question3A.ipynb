```python
edge_betweenness = nx.edge_betweenness_centrality(G)
```

```python
import heapq

heap = [(-score, edge) for edge, score in edge_betweenness.items()]
```

```python
remove_n_edges = 15

while remove_n_edges>0:
    heapq.heapify(heap)
    curr_top_score, curr_top_edge = heapq.heappop(heap)
    G.remove_edge(curr_top_edge[0], curr_top_edge[1])
    remove_n_edges -= 1
    
```
